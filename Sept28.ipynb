```python
###comparison operations

jay= 21
viru=22
""" who will pay the bill ?"""
if jay>viru:
    print("jay and viru will pay the bill")
else:
    print("viru will pay the bill")  
print("Greater than:",jay>viru)

print("Greater than or equal to:",jay>=viru)
print("Less than or equal to:",jay<=viru)
print("Equal:",jay==viru)
print("Not Equal:",jay!=viru)
```

```python
#### ARITHMATIC OPERATIONS

N1=100
N2=20
print("Addition:",N1+N2)
print("Subtraction:",N1-N2)
print("Multiplication:",N1*N2)
print("Division:",N1/N2)
print("Floor Division:",N1//N2)
```

```python
### COMPARISON OPERATIONS
monu=24
sonu=22
print("Equal:",monu==sonu)
print("Not Equal:",monu!=sonu)
print("Greater than:",monu>sonu)
print("Less than:",monu<sonu)
print("Greater than or equal to:",monu>=sonu)
print("Less than or equal to:",monu<=sonu)
print("Comparison Operations:",monu==sonu,monu!=sonu,monu>sonu,monu<sonu,monu>=sonu,monu<=sonu)
```

```python
##logical operations
a=True
b=False
print("AND operation:",a and b)
print("OR operation:",a or b)
print("NOT operation:",not a)
```
